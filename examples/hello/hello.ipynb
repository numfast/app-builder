# Hello App Demo

Два варианта одного приложения, собранных App Builder'ом:
1. **Hello** — _main + Greeting (`full.toml`)
2. **Hello+Bondiana** — _main + Greeting + Bondiana (`full.bondiana.toml`)

Bondiana переопределяет `_default`: последний зарегистрированный алиас побеждает.

```python
import sys
sys.path.insert(0, r"C:\App\numfast\app-builder\examples\hello\dist")

# Импортируем оба собранных приложения
from helloapp import APP as hello_app
from full_bondiana import APP as bondiana_app

print("=== Hello App (plain) ===")
print("Name:", hello_app.metadata["_kernel"]["name"])
print("Aliases:", [a for a in hello_app.alias.keys() if not a.startswith("_")])
print()
print("=== Hello+Bondiana App ===")
print("Name:", bondiana_app.metadata["_kernel"]["name"])
print("Aliases:", [a for a in bondiana_app.alias.keys() if not a.startswith("_")])
```

## Вызов _default

У обоих приложений есть `_default`. Но Bondiana его переопределяет.

```python
print("Hello._default:")
print("  ", hello_app)
print("  function:", hello_app.alias["_default"].__name__)
print()
print("Hello+Bondiana._default:")
print("  ", bondiana_app)
print("  function:", bondiana_app.alias["_default"].__name__)
```

## Вызов hello

Функция `hello` определена в Greeting и доступна в обоих приложениях.

```python
print("Hello.hello('World'):", hello_app.call("hello", "World"))
print("Bondiana.hello('World'):", bondiana_app.call("hello", "World"))
```

## Системные алиасы (namespaced)

Каждое расширение регистрирует также алиас вида `_ExtensionName_alias`.
Это позволяет вызывать оригинальную функцию даже если алиас переопределён.

```python
print("Bondiana: все алиасы (включая системные):")
for name in sorted(bondiana_app.alias.keys()):
    fn = bondiana_app.alias[name]
    print(f"  {name:30s} -> {fn.__name__}")
```

## Как это работает

1. Builder читает `full.toml`, находит расширения
2. Сортирует по зависимостям (сохраняя порядок из манифеста)
3. Склеивает `.py` файлы в один
4. Добавляет `Kernel` и регистрацию алиасов
5. Последний зарегистрированный алиас побеждает

Bondiana идёт ПОСЛЕ _main и Greeting, поэтому её `_default` переопределяет предыдущий.
