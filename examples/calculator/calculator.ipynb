# Calculator Demo

Примеры использования калькулятора, собранного App Builder'ом.

Сборка: `MAIN["build"]("examples/calculator")`

```python
import sys
sys.path.insert(0, r"C:\App\numfast\app-builder")

from builder import MAIN

# Собираем калькулятор
calc = MAIN["build"](r"C:\App\numfast\app-builder\examples\calculator")
print("Calculator built:", calc.metadata["_kernel"]["name"])
print("Aliases:", [k for k in calc.alias.keys() if not k.startswith("_")])
```

## Арифметика

```python
print("add(2, 3) =", calc.alias["add"](2, 3))
print("sub(10, 4) =", calc.alias["sub"](10, 4))
print("mul(3, 5) =", calc.alias["mul"](3, 5))
print("div(9, 4) =", calc.alias["div"](9, 4))
print("pow(2, 10) =", calc.alias["pow"](2, 10))
```

## Тригонометрия

```python
print("sin(0) =", calc.alias["sin"](0))
print("cos(0) =", calc.alias["cos"](0))
print("tan(0) =", calc.alias["tan"](0))
print("sin(pi/2) =", calc.alias["sin"](calc.alias["pi"]() / 2))
print("pi =", calc.alias["pi"]())
```

## Калькулятор выражений

```python
print('calc("2 + 2 * 3") =', calc.alias["calc"]("2 + 2 * 3"))
print('calc("(3 + 5) * 2") =', calc.alias["calc"]("(3 + 5) * 2"))
print('calc("pow(2, 3)") =', calc.alias["calc"]("pow(2, 3)"))
```

## Пакетный расчёт

```python
import random

print("10 random calculations:\n")
ops = ["add", "sub", "mul", "div", "pow", "sin", "cos", "tan"]
for i in range(10):
    op = random.choice(ops)
    if op in ("sin", "cos", "tan"):
        x = random.uniform(0, 3.14)
        result = calc.alias[op](x)
        print(f"  {op}({x:.3f}) = {result}")
    else:
        a = random.randint(1, 20)
        b = random.randint(1, 10)
        if op == "div":
            b = random.randint(1, 10)
        result = calc.alias[op](a, b)
        print(f"  {op}({a}, {b}) = {result}")
```

## Тесты из манифеста

```python
tests = calc.metadata.get("_tests", {})
print(f"Running {len(tests)} tests from manifest...\n")
passed = 0
for name, spec in tests.items():
    fn = calc.alias.get(spec["mod"])
    if fn is None:
        print(f"  FAIL {name}: alias not found")
        continue
    result = str(fn(*spec["args"]))
    expected = spec["expected"]
    ok = result == expected
    status = "OK" if ok else f"FAIL (got {result})"
    print(f"  {status}: {name}")
    if ok:
        passed += 1
print(f"\nResult: {passed}/{len(tests)}")
```
